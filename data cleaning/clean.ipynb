## Removing small and grayscale images

```python
def checkImage(image:str, path:str):
    image = cv2.imread(os.path.join(path, image))

    if image.shape[0] < 512 or image.shape[1] < 512:
        return False
    
    hsl = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
    saturation = np.array(hsl)[:, :, 1].mean()

    return saturation > 10
```
